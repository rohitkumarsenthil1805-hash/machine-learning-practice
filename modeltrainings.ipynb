```python
import pandas as pd
import numpy as np 
```

```python
df=pd.read_csv("BankNote_Authentication.csv ")
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
x=df.drop(["class"], axis=1)
```

```python
y=df["class"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=50,test_size=0.25)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report,mean_squared_error,r2_score,mean_absolute_error
```

```python
def metrics(true,prediction):
    print("Accuracy Score:",accuracy_score(true,prediction))
    print("Confusion Matrix:")
    print(confusion_matrix(true,prediction))
    print("Classification Report:")
    print(classification_report(true,prediction))
    print("Mean Squared Error:",mean_squared_error(true,prediction))
    print("R-squared Score:",r2_score(true,prediction))
    print("Mean Absolute Error:",mean_absolute_error(true,prediction))
    
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
classifier=RandomForestClassifier()
```

```python
classifier.fit(x_train,y_train)
y_pred=classifier.predict(x_test)
```

```python
metrics(y_test,y_pred)
```

```python
import pickle
```

```python
po=open("classifier.pkl","wb")  
pickle.dump(classifier,po)
po.close()
```
